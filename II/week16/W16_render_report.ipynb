# AMPS II · Week 16 — Rendering the final report on Colab

**Render `report.qmd` → `report.html`** with Quarto (about 1 minute for the download, once per runtime). Skip this notebook if you have Quarto on your own computer:
`quarto render report.qmd` in the report folder. The same cell renders Quarto slides (e.g. `progress_talk.qmd` from Week 15, adapted for the final talk).

The report itself contains no code cells: all numbers and figures come from your `project.ipynb`. No DOLFINx is needed here.

## 1 · Your report folder
Put `report.qmd`, `references.bib` and `figs/` in a folder on Google Drive (e.g. `MyDrive/AMPS2_project`) and set the path below.

```python
import os
from google.colab import drive
drive.mount("/content/drive")
REPORT_DIR = "/content/drive/MyDrive/AMPS2_project"      # <- change if needed
print(sorted(os.listdir(REPORT_DIR)))
```

## 2 · Install Quarto (once per runtime) and render

```python
import os, subprocess
V = "1.9.36"
QUARTO = f"/content/quarto-{V}/bin/quarto"
if not os.path.exists(QUARTO):
    url = f"https://github.com/quarto-dev/quarto-cli/releases/download/v{V}/quarto-{V}-linux-amd64.tar.gz"
    subprocess.run(["wget", "-q", "-O", "/content/quarto.tar.gz", url], check=True)
    subprocess.run(["tar", "-xzf", "/content/quarto.tar.gz", "-C", "/content"], check=True)
print(subprocess.run([QUARTO, "--version"], capture_output=True, text=True).stdout.strip())
```

```python
DOC = "report.qmd"                                      # or your slides, e.g. "final_talk.qmd"
r = subprocess.run([QUARTO, "render", DOC], cwd=REPORT_DIR, capture_output=True, text=True)
print(r.stdout[-1500:], r.stderr[-3000:])
out = os.path.join(REPORT_DIR, DOC.replace(".qmd", ".html"))
print(out, "written:", os.path.exists(out))
```

Open the HTML file from Drive (download it) and check that the figures appear and the equations render. With `embed-resources: true` (in the template) the
HTML file is self-contained — that single file is what you submit together with `report.qmd`.
